```python
%sql

CREATE OR REPLACE TABLE workspace.gridpulse_gold.fact_meter_reading AS
SELECT
    c.reading_id,
    c.meter_id,
    m.building_id,
    DATE(c.reading_ts) AS date_key,
    c.reading_ts,
    c.energy_kwh,
    c.active_power_kw,
    c.voltage_v,
    c.current_a,
    c.power_factor,
    m.tariff_plan_id,
    t.tariff_id,
    t.time_band,
    t.rate_per_kwh,
    t.peak_flag,
    c.energy_kwh * t.rate_per_kwh AS estimated_cost,
    c.reading_quality_flag,
    c.source_system,
    c.producer_run_id,
    c.reading_failed,
    c.dq_status
FROM workspace.gridpulse_silver.consumption_trusted c
JOIN workspace.gridpulse_gold.dim_meter m
    ON c.meter_id = m.meter_id
LEFT JOIN workspace.gridpulse_gold.dim_tariff t
    ON m.tariff_plan_id = t.tariff_plan_id
    AND HOUR(c.reading_ts) >= t.start_hour
    AND HOUR(c.reading_ts) < t.end_hour
    AND c.reading_ts >= t.effective_from
    AND c.reading_ts <= t.effective_to
WHERE c.dq_status = 'PASS';
```

```python
%sql

SELECT *
FROM workspace.gridpulse_gold.fact_meter_reading
ORDER BY reading_ts, meter_id
LIMIT 20;
```

```python
%sql

CREATE OR REPLACE TABLE workspace.gridpulse_gold.agg_meter_health_daily AS
SELECT
    meter_id,
    building_id,
    date_key,
    COUNT(*) AS reading_count,
    AVG(power_factor) AS avg_power_factor,
    MIN(reading_ts) AS first_reading_ts,
    MAX(reading_ts) AS last_reading_ts,
    SUM(CASE WHEN reading_quality_flag = 'ESTIMATED' THEN 1 ELSE 0 END) AS estimated_reading_count,
    SUM(CASE WHEN reading_quality_flag = 'ACTUAL' THEN 1 ELSE 0 END) AS actual_reading_count
FROM workspace.gridpulse_gold.fact_meter_reading
WHERE dq_status = 'PASS'
GROUP BY
    meter_id,
    building_id,
    date_key;
```

```python
%sql

CREATE OR REPLACE TABLE workspace.gridpulse_gold.agg_campus_anomaly_daily AS
SELECT
    date_key,
    COUNT(*) AS total_readings,
    SUM(
        CASE
            WHEN reading_quality_flag <> 'ACTUAL'
            THEN 1
            ELSE 0
        END
    ) AS anomaly_reading_count,
    ROUND(
        SUM(
            CASE
                WHEN reading_quality_flag <> 'ACTUAL'
                THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        4
    ) AS anomaly_rate
FROM workspace.gridpulse_gold.fact_meter_reading
WHERE dq_status = 'PASS'
GROUP BY date_key;
```

```python
%sql

SELECT *
FROM workspace.gridpulse_gold.agg_campus_anomaly_daily
ORDER BY date_key;
```
